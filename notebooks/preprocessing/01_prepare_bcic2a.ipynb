# BCIC IV-2a preprocessing

이 notebook은 전처리 구현을 복제하지 않고 프로젝트 CLI를 실행합니다. `PROTOCOL`만 바꾸면 canonical MI-9 또는 channel-overlap 데이터를 생성할 수 있습니다. Session 1은 train/validation에만, Session 2는 final test에만 사용됩니다.

```python
from pathlib import Path
import json
import subprocess
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / 'scripts' / 'preprocess_bcic2a.py').exists():
    PROJECT_ROOT = PROJECT_ROOT.parents[1]

PROTOCOL = 'channel_overlap'  # or 'canonical_mi9'
CONFIG_PATH = PROJECT_ROOT / 'configs' / 'preprocessing' / f'{PROTOCOL}.yaml'
OUTPUT_DIR = PROJECT_ROOT / 'artifacts' / 'preprocessed' / 'bcic2a' / PROTOCOL
CONFIG_PATH, OUTPUT_DIR
```

```python
command = [
    sys.executable,
    str(PROJECT_ROOT / 'scripts' / 'preprocess_bcic2a.py'),
    '--config',
    str(CONFIG_PATH),
]
subprocess.run(command, cwd=PROJECT_ROOT, check=True)
```

```python
summary_path = OUTPUT_DIR / 'preprocessing_summary.json'
summary = json.loads(summary_path.read_text(encoding='utf-8'))
summary['split_shapes']
```
